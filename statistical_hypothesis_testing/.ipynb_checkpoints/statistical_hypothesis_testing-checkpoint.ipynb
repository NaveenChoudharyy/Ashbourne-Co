# STATISTICAL HYPOTHESIS TESTING
============================================================

### BUSINESS CONTEXT:
The executive board has formulated several operational and
commercial assumptions based on observed business patterns.

The objective of this analysis is to statistically validate these
assumptions using appropriate hypothesis tests in Python.

Rather than relying only on descriptive statistics and visual
patterns, statistical hypothesis testing will determine whether
observed differences or relationships are statistically significant
at the specified significance levels.

For each hypothesis, the analysis should clearly define:

• Null Hypothesis (H₀)
• Alternative Hypothesis (H₁)
• Significance Level (α)
• Appropriate Statistical Test
• Test Statistic
• P-value
• Decision regarding H₀
• Business Interpretation

The statistical conclusions should be based on the selected
significance level for each individual hypothesis.

### 1. RETURN RATE CRISIS — TWO-PROPORTION Z-TEST

##### Is the proportion of returned orders statistically greater for E-Commerce than for physical stores?

```python
# H0: E-Commerce return proportion <= Physical-store return proportion
# H1: E-Commerce return proportion > Physical-store return proportion

# Significance Level: α = 0.01


# Make sure return_status is treated consistently
orders["return_status"] = orders["return_status"].str.strip()

# Identify E-Commerce stores
ecommerce_store_ids = stores.loc[
    stores["store_type"].str.strip().str.lower() == "e-commerce",
    "store_id"
]

# Create store type for every order
orders_test = orders.copy()

orders_test["store_group"] = np.where(
    orders_test["store_id"].isin(ecommerce_store_ids),
    "E-Commerce",
    "Physical Store"
)

# -------------------------------------------------------------------------
# Calculate total and returned orders
# -------------------------------------------------------------------------

return_summary = (
    orders_test
    .groupby("store_group")
    .agg(
        total_orders=("order_id", "nunique"),
        returned_orders=(
            "return_status",
            lambda x: (x.str.lower() == "returned").sum()
        )
    )
    .reset_index()
)

# Calculate return rate
return_summary["return_rate"] = (
    return_summary["returned_orders"] /
    return_summary["total_orders"]
)

print("=" * 70)
print("RETURN RATE — E-COMMERCE VS PHYSICAL STORES")
print("=" * 70)

display(
    return_summary.assign(
        return_rate=lambda x: x["return_rate"] * 100
    ).round(2)
)

# -------------------------------------------------------------------------
# Extract values for hypothesis test
# -------------------------------------------------------------------------

ecommerce_data = return_summary[
    return_summary["store_group"] == "E-Commerce"
].iloc[0]

physical_data = return_summary[
    return_summary["store_group"] == "Physical Store"
].iloc[0]

# Number of returned orders
count = np.array([
    ecommerce_data["returned_orders"],
    physical_data["returned_orders"]
])

# Total number of orders
nobs = np.array([
    ecommerce_data["total_orders"],
    physical_data["total_orders"]
])

# -------------------------------------------------------------------------
# TWO-PROPORTION Z-TEST
#
# Alternative hypothesis:
# E-Commerce return rate > Physical-store return rate
#
# Therefore, use a one-sided ("larger") test.
# -------------------------------------------------------------------------

z_stat, p_value = proportions_ztest(
    count=count,
    nobs=nobs,
    alternative="larger"
)

alpha = 0.01

# -------------------------------------------------------------------------
# Calculate return rates
# -------------------------------------------------------------------------

ecommerce_return_rate = (
    ecommerce_data["returned_orders"] /
    ecommerce_data["total_orders"]
)

physical_return_rate = (
    physical_data["returned_orders"] /
    physical_data["total_orders"]
)

# -------------------------------------------------------------------------
# Calculate difference in return rates
# -------------------------------------------------------------------------

rate_difference = (
    ecommerce_return_rate -
    physical_return_rate
)

# -------------------------------------------------------------------------
# Hypothesis decision
# -------------------------------------------------------------------------

if p_value < alpha:
    decision = "Reject H₀"
    conclusion = (
        "There is statistically significant evidence at the 1% "
        "significance level that the E-Commerce return rate is "
        "greater than the physical-store return rate."
    )
else:
    decision = "Fail to Reject H₀"
    conclusion = (
        "There is insufficient statistical evidence at the 1% "
        "significance level to conclude that the E-Commerce return "
        "rate is greater than the physical-store return rate."
    )

# -------------------------------------------------------------------------
# DISPLAY RESULTS
# -------------------------------------------------------------------------

print("\n" + "=" * 70)
print("HYPOTHESIS TEST RESULTS")
print("=" * 70)

print(f"\nSignificance Level (α): {alpha}")

print(
    f"\nE-Commerce Return Rate: "
    f"{ecommerce_return_rate * 100:.2f}%"
)

print(
    f"Physical Store Return Rate: "
    f"{physical_return_rate * 100:.2f}%"
)

print(
    f"Difference in Return Rates: "
    f"{rate_difference * 100:.2f} percentage points"
)

print(f"\nZ-Statistic: {z_stat:.4f}")
print(f"P-Value: {p_value:.6f}")

print(f"\nDecision: {decision}")

print("\nBusiness Interpretation:")
print(conclusion)

# -------------------------------------------------------------------------
# OPTIONAL VISUAL COMPARISON
# -------------------------------------------------------------------------

plot_data = pd.DataFrame({
    "Store Type": [
        "E-Commerce",
        "Physical Store"
    ],
    "Return Rate (%)": [
        ecommerce_return_rate * 100,
        physical_return_rate * 100
    ]
})

plt.figure(figsize=(9, 6))

sns.barplot(
    data=plot_data,
    x="Store Type",
    y="Return Rate (%)"
)

plt.title(
    "Return Rate: E-Commerce vs Physical Stores",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Store Type")
plt.ylabel("Return Rate (%)")

plt.tight_layout()
plt.show()
```

### 2. GENDER-BASED AD TARGETING — TWO-SAMPLE HYPOTHESIS TEST

##### Is there a statistically significant difference in Average Order Value(AOV) between Male and Female shoppers?

```python
# H0: There is no statistically significant difference in AOV
#     between Male and Female shoppers.

# H1: There is a statistically significant difference in AOV
#     between Male and Female shoppers.

# Significance Level: α = 0.05
# -------------------------------------------------------------------------

# Merge customer gender with orders
gender_orders = orders.merge(
    customers[["customer_id", "gender"]],
    on="customer_id",
    how="left"
)

# Keep only Male and Female customers
gender_orders = gender_orders[
    gender_orders["gender"].isin(["Male", "Female"])
].copy()

# -------------------------------------------------------------------------
# Calculate order-level AOV
# -------------------------------------------------------------------------
# If an order contains multiple rows, aggregate them first so that each
# order contributes one AOV observation.

order_aov = (
    gender_orders
    .groupby(
        ["order_id", "gender"],
        as_index=False
    )["net_sales"]
    .sum()
)

order_aov = order_aov.rename(
    columns={"net_sales": "aov"}
)

# Separate Male and Female AOV
male_aov = order_aov.loc[
    order_aov["gender"] == "Male",
    "aov"
].dropna()

female_aov = order_aov.loc[
    order_aov["gender"] == "Female",
    "aov"
].dropna()

# -------------------------------------------------------------------------
# Descriptive statistics
# -------------------------------------------------------------------------

print("=" * 70)
print("AVERAGE ORDER VALUE — MALE VS FEMALE")
print("=" * 70)

print(f"\nMale Orders: {len(male_aov):,}")
print(f"Female Orders: {len(female_aov):,}")

print(f"\nMale Average Order Value: £{male_aov.mean():,.2f}")
print(f"Female Average Order Value: £{female_aov.mean():,.2f}")

print(f"\nMale AOV Standard Deviation: £{male_aov.std():,.2f}")
print(f"Female AOV Standard Deviation: £{female_aov.std():,.2f}")

# -------------------------------------------------------------------------
# TEST SELECTION
# -------------------------------------------------------------------------
# Welch's independent two-sample t-test is used because it does not assume
# equal variance between Male and Female AOV distributions.
#
# Two-sided test because H1 states that the AOVs are different, without
# specifying which gender must have the higher AOV.
# -------------------------------------------------------------------------

t_stat, p_value = stats.ttest_ind(
    male_aov,
    female_aov,
    equal_var=False,
    alternative="two-sided"
)

alpha = 0.05

# -------------------------------------------------------------------------
# Difference in mean AOV
# -------------------------------------------------------------------------

mean_difference = (
    male_aov.mean() -
    female_aov.mean()
)

# -------------------------------------------------------------------------
# Hypothesis decision
# -------------------------------------------------------------------------

if p_value < alpha:
    decision = "Reject H₀"
    
    conclusion = (
        "There is statistically significant evidence at the 5% "
        "significance level that Average Order Value differs between "
        "Male and Female shoppers."
    )
else:
    decision = "Fail to Reject H₀"
    
    conclusion = (
        "There is insufficient statistical evidence at the 5% "
        "significance level to conclude that Average Order Value "
        "differs between Male and Female shoppers."
    )

# -------------------------------------------------------------------------
# RESULTS
# -------------------------------------------------------------------------

print("\n" + "=" * 70)
print("HYPOTHESIS TEST RESULTS")
print("=" * 70)

print("\nH₀: Male AOV = Female AOV")
print("H₁: Male AOV ≠ Female AOV")

print(f"\nSignificance Level (α): {alpha}")

print(f"\nMale Mean AOV: £{male_aov.mean():,.2f}")
print(f"Female Mean AOV: £{female_aov.mean():,.2f}")

print(f"\nDifference in Mean AOV: £{mean_difference:,.2f}")

print(f"\nT-Statistic: {t_stat:.4f}")
print(f"P-Value: {p_value:.6f}")

print(f"\nDecision: {decision}")

print("\nBusiness Interpretation:")
print(conclusion)


# =============================================================================
# VISUAL COMPARISON
# =============================================================================

plot_data = pd.DataFrame({
    "Gender": ["Male", "Female"],
    "Average_AOV": [
        male_aov.mean(),
        female_aov.mean()
    ],
    "Std_AOV": [
        male_aov.std(),
        female_aov.std()
    ]
})

plt.figure(figsize=(9, 6))

plt.bar(
    plot_data["Gender"],
    plot_data["Average_AOV"],
    yerr=plot_data["Std_AOV"],
    capsize=5
)

plt.title(
    "Average Order Value by Gender",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Gender")
plt.ylabel("Average Order Value (£)")

plt.tight_layout()
plt.show()
```

### 3. INVENTORY SEASONALITY — CHI-SQUARE TEST OF INDEPENDENCE

##### Is Product Category statistically independent of Purchasing Season?

```python
# H0: Product Category and Purchasing Season are independent.
# H1: Product Category and Purchasing Season are not independent.

# Significance Level: α = 0.025
# -------------------------------------------------------------------------

# Merge product category and season with orders
season_data = orders.merge(
    products[
        [
            "product_id",
            "category",
            "season"
        ]
    ],
    on="product_id",
    how="left"
)

# Keep only the two required seasons
season_data = season_data[
    season_data["season"].isin([
        "Autumn/Winter",
        "Spring/Summer"
    ])
].copy()

# Remove missing categories/seasons
season_data = season_data.dropna(
    subset=["category", "season"]
)

# -------------------------------------------------------------------------
# Create contingency table
# -------------------------------------------------------------------------

contingency_table = pd.crosstab(
    season_data["category"],
    season_data["season"]
)

print("=" * 70)
print("PRODUCT CATEGORY vs PURCHASING SEASON")
print("=" * 70)

print("\nObserved Frequency Table:")
display(contingency_table)

# -------------------------------------------------------------------------
# Chi-Square Test of Independence
# -------------------------------------------------------------------------

chi2_stat, p_value, degrees_freedom, expected = chi2_contingency(
    contingency_table
)

alpha = 0.025

# -------------------------------------------------------------------------
# Expected frequency table
# -------------------------------------------------------------------------

expected_table = pd.DataFrame(
    expected,
    index=contingency_table.index,
    columns=contingency_table.columns
)

print("\nExpected Frequency Table:")
display(expected_table.round(2))

# -------------------------------------------------------------------------
# Hypothesis decision
# -------------------------------------------------------------------------

if p_value < alpha:

    decision = "Reject H₀"

    conclusion = (
        "There is statistically significant evidence at the 2.5% "
        "significance level that Product Category and Purchasing "
        "Season are not independent."
    )

else:

    decision = "Fail to Reject H₀"

    conclusion = (
        "There is insufficient statistical evidence at the 2.5% "
        "significance level to conclude that Product Category and "
        "Purchasing Season are associated."
    )

# -------------------------------------------------------------------------
# RESULTS
# -------------------------------------------------------------------------

print("\n" + "=" * 70)
print("CHI-SQUARE HYPOTHESIS TEST RESULTS")
print("=" * 70)

print("\nH₀: Product Category and Purchasing Season are independent.")
print("H₁: Product Category and Purchasing Season are not independent.")

print(f"\nSignificance Level (α): {alpha}")

print(f"\nChi-Square Statistic: {chi2_stat:.4f}")
print(f"Degrees of Freedom: {degrees_freedom}")
print(f"P-Value: {p_value:.6f}")

print(f"\nDecision: {decision}")

print("\nBusiness Interpretation:")
print(conclusion)


# =============================================================================
# IDENTIFY CATEGORY-SEASON COMBINATIONS CONTRIBUTING TO THE RESULT
# =============================================================================

# Calculate standardized residuals
# Large positive values indicate more observations than expected.
# Large negative values indicate fewer observations than expected.

standardized_residuals = (
    contingency_table - expected_table
) / np.sqrt(expected_table)

print("\nStandardized Residuals:")
display(standardized_residuals.round(2))


# =============================================================================
# VISUALIZATION — PURCHASING SEASON BY PRODUCT CATEGORY
# =============================================================================

contingency_table.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title(
    "Purchasing Season by Product Category",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Product Category")
plt.ylabel("Number of Orders")
plt.xticks(rotation=0)
plt.legend(title="Season")

plt.tight_layout()
plt.show()


# =============================================================================
# OPTIONAL — PROPORTIONAL VIEW
# =============================================================================

# Convert counts into percentages within each product category
category_season_pct = (
    contingency_table
    .div(contingency_table.sum(axis=1), axis=0)
    * 100
)

category_season_pct.plot(
    kind="bar",
    stacked=True,
    figsize=(12, 6)
)

plt.title(
    "Seasonal Purchasing Mix by Product Category",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Product Category")
plt.ylabel("Percentage of Orders (%)")
plt.xticks(rotation=0)
plt.legend(title="Season")

plt.tight_layout()
plt.show()
```

### 4. FORMAT MARGIN DISCREPANCY — ONE-WAY ANOVA

##### Do average profit margins significantly differ across the three physical store formats: Outlet, Boutique, and Flagship?

```python
# H0: Mean profit margin is equal across all three physical store formats.
# H1: At least one physical store format has a different mean profit margin.

# Significance Level: α = 0.10
# -------------------------------------------------------------------------

# Merge store format information with orders
margin_data = orders.merge(
    stores[
        [
            "store_id",
            "store_type"
        ]
    ],
    on="store_id",
    how="left"
)

# -------------------------------------------------------------------------
# Keep only the three physical store formats
# -------------------------------------------------------------------------

physical_formats = [
    "Outlet",
    "Boutique",
    "Flagship"
]

margin_data = margin_data[
    margin_data["store_type"].isin(physical_formats)
].copy()

# -------------------------------------------------------------------------
# Calculate order-level profit margin
#
# Profit Margin = Profit / Net Sales × 100
# -------------------------------------------------------------------------

margin_data["profit_margin"] = np.where(
    margin_data["net_sales"] != 0,
    (margin_data["profit"] / margin_data["net_sales"]) * 100,
    np.nan
)

# Remove invalid/missing margins
margin_data = margin_data.dropna(
    subset=["profit_margin"]
)

# -------------------------------------------------------------------------
# Descriptive statistics by store format
# -------------------------------------------------------------------------

margin_summary = (
    margin_data
    .groupby("store_type")["profit_margin"]
    .agg(
        Average_Margin="mean",
        Median_Margin="median",
        Std_Deviation="std",
        Order_Count="count"
    )
    .reset_index()
)

print("=" * 70)
print("PROFIT MARGIN BY PHYSICAL STORE FORMAT")
print("=" * 70)

display(
    margin_summary.round(2)
)

# -------------------------------------------------------------------------
# Create separate groups for ANOVA
# -------------------------------------------------------------------------

outlet_margin = margin_data.loc[
    margin_data["store_type"] == "Outlet",
    "profit_margin"
]

boutique_margin = margin_data.loc[
    margin_data["store_type"] == "Boutique",
    "profit_margin"
]

flagship_margin = margin_data.loc[
    margin_data["store_type"] == "Flagship",
    "profit_margin"
]

# -------------------------------------------------------------------------
# ONE-WAY ANOVA
# -------------------------------------------------------------------------

f_stat, p_value = stats.f_oneway(
    outlet_margin,
    boutique_margin,
    flagship_margin
)

alpha = 0.10

# -------------------------------------------------------------------------
# Hypothesis decision
# -------------------------------------------------------------------------

if p_value < alpha:

    decision = "Reject H₀"

    conclusion = (
        "There is statistically significant evidence at the 10% "
        "significance level that average profit margins differ "
        "across the physical store formats."
    )

else:

    decision = "Fail to Reject H₀"

    conclusion = (
        "There is insufficient statistical evidence at the 10% "
        "significance level to conclude that average profit margins "
        "differ across the physical store formats."
    )

# -------------------------------------------------------------------------
# RESULTS
# -------------------------------------------------------------------------

print("\n" + "=" * 70)
print("ONE-WAY ANOVA RESULTS")
print("=" * 70)

print("\nH₀: Mean profit margin is equal across all three store formats.")
print("H₁: At least one store format has a different mean profit margin.")

print(f"\nSignificance Level (α): {alpha}")

print(f"\nF-Statistic: {f_stat:.4f}")
print(f"P-Value: {p_value:.6f}")

print(f"\nDecision: {decision}")

print("\nBusiness Interpretation:")
print(conclusion)


# =============================================================================
# VISUALIZATION — PROFIT MARGIN BY STORE FORMAT
# =============================================================================

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=margin_data,
    x="store_type",
    y="profit_margin"
)

plt.title(
    "Profit Margin Distribution by Physical Store Format",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Store Format")
plt.ylabel("Profit Margin (%)")

plt.tight_layout()
plt.show()


# =============================================================================
# POST-HOC ANALYSIS
# =============================================================================
# ANOVA only tells us whether at least one group differs.
# If the ANOVA is significant, Tukey's HSD identifies which specific
# store formats differ from one another.
# =============================================================================

if p_value < alpha:

    from statsmodels.stats.multicomp import pairwise_tukeyhsd

    tukey_result = pairwise_tukeyhsd(
        endog=margin_data["profit_margin"],
        groups=margin_data["store_type"],
        alpha=alpha
    )

    print("\n" + "=" * 70)
    print("TUKEY HSD POST-HOC TEST")
    print("=" * 70)

    print(tukey_result)

else:

    print("\nTukey HSD was not performed because the overall ANOVA")
    print("was not statistically significant at α = 0.10.")
```

####
### 5. DISCOUNT CANNIBALIZATION — PEARSON CORRELATION TEST

##### Is there a statistically significant linear relationship between Discount Percentage and Quantity Purchased per Order?

```python
# H0: There is no linear correlation between Discount Percentage and Quantity Purchased per Order.
# H1: There is a statistically significant linear correlation between Discount Percentage and Quantity Purchased per Order.

# Significance Level: α = 0.01
# -------------------------------------------------------------------------

# Select required columns
discount_data = orders[
    [
        "order_id",
        "discount_pct",
        "quantity"
    ]
].copy()

# Remove missing values
discount_data = discount_data.dropna(
    subset=[
        "discount_pct",
        "quantity"
    ]
)

# -------------------------------------------------------------------------
# Pearson Correlation Test
# -------------------------------------------------------------------------

correlation, p_value = stats.pearsonr(
    discount_data["discount_pct"],
    discount_data["quantity"]
)

alpha = 0.01

# -------------------------------------------------------------------------
# Interpret correlation strength
# -------------------------------------------------------------------------

abs_correlation = abs(correlation)

if abs_correlation < 0.20:
    strength = "Very weak"

elif abs_correlation < 0.40:
    strength = "Weak"

elif abs_correlation < 0.60:
    strength = "Moderate"

elif abs_correlation < 0.80:
    strength = "Strong"

else:
    strength = "Very strong"


if correlation > 0:
    direction = "positive"
elif correlation < 0:
    direction = "negative"
else:
    direction = "no"


# -------------------------------------------------------------------------
# Hypothesis decision
# -------------------------------------------------------------------------

if p_value < alpha:

    decision = "Reject H₀"

    conclusion = (
        "There is statistically significant evidence at the 1% "
        "significance level of a linear relationship between "
        "Discount Percentage and Quantity Purchased per Order."
    )

else:

    decision = "Fail to Reject H₀"

    conclusion = (
        "There is insufficient statistical evidence at the 1% "
        "significance level to conclude that Discount Percentage "
        "and Quantity Purchased per Order have a linear relationship."
    )


# =============================================================================
# RESULTS
# =============================================================================

print("=" * 70)
print("DISCOUNT CANNIBALIZATION — PEARSON CORRELATION TEST")
print("=" * 70)

print("\nH₀: There is no linear correlation between discount percentage")
print("    and quantity purchased per order.")

print("\nH₁: There is a statistically significant linear correlation")
print("    between discount percentage and quantity purchased per order.")

print(f"\nSignificance Level (α): {alpha}")

print(f"\nNumber of Orders Tested: {len(discount_data):,}")

print(f"\nPearson Correlation Coefficient (r): {correlation:.4f}")

print(f"Correlation Strength: {strength}")
print(f"Correlation Direction: {direction}")

print(f"\nP-Value: {p_value:.6f}")

print(f"\nDecision: {decision}")

print("\nBusiness Interpretation:")
print(conclusion)


# =============================================================================
# SCATTER PLOT WITH REGRESSION LINE
# =============================================================================

plt.figure(figsize=(12, 7))

sns.regplot(
    data=discount_data,
    x="discount_pct",
    y="quantity",
    scatter_kws={"alpha": 0.4, "s": 40},
    line_kws={"linewidth": 2}
)

plt.title(
    "Discount Percentage vs Quantity Purchased",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Discount Percentage")
plt.ylabel("Quantity Purchased per Order")

plt.tight_layout()
plt.show()


# =============================================================================
# ADDITIONAL BUSINESS METRICS
# =============================================================================

# Compare average quantity across discount bands
discount_data["discount_band"] = pd.cut(
    discount_data["discount_pct"],
    bins=[
        -np.inf,
        0,
        0.10,
        0.20,
        0.30,
        0.40,
        np.inf
    ],
    labels=[
        "0%",
        "1–10%",
        "11–20%",
        "21–30%",
        "31–40%",
        "40%+"
    ]
)

discount_band_summary = (
    discount_data
    .groupby(
        "discount_band",
        observed=False
    )["quantity"]
    .agg(
        Average_Quantity="mean",
        Median_Quantity="median",
        Order_Count="count"
    )
    .reset_index()
)

print("\n" + "=" * 70)
print("QUANTITY BY DISCOUNT BAND")
print("=" * 70)

display(
    discount_band_summary.round(2)
)
```

### 6. QUALITY EXPECTATIONS — SPEARMAN RANK CORRELATION

##### Is there a statistically significant monotonic relationship between customer Income Bracket and the 1–5 star Rating assigned to orders?

```python
# H0: There is no statistically significant monotonic relationship between Income Bracket and Rating.
# H1: There is a statistically significant monotonic relationship between Income Bracket and Rating.

# Significance Level: α = 0.05
# -------------------------------------------------------------------------

# Merge customer income bracket with order ratings
quality_data = orders.merge(
    customers[
        [
            "customer_id",
            "income_bracket"
        ]
    ],
    on="customer_id",
    how="left"
)

# Keep only records with valid income bracket and rating
quality_data = quality_data.dropna(
    subset=[
        "income_bracket",
        "rating"
    ]
).copy()

# Keep valid 1–5 star ratings
quality_data = quality_data[
    quality_data["rating"].between(1, 5)
].copy()

# -------------------------------------------------------------------------
# Inspect the actual income bracket values
# -------------------------------------------------------------------------

print("=" * 70)
print("INCOME BRACKET VALUES")
print("=" * 70)

print(
    quality_data["income_bracket"]
    .value_counts()
)

# -------------------------------------------------------------------------
# Convert Income Bracket into an ordinal numeric scale
#
# IMPORTANT:
# Update this mapping if your dataset uses different income-bracket
# labels or a different natural ordering.
# -------------------------------------------------------------------------

income_order = {
    "Low": 1,
    "Medium": 2,
    "High": 3,
    "Ultra-High": 4
}

quality_data["income_score"] = (
    quality_data["income_bracket"]
    .map(income_order)
)

# Remove values that were not present in the mapping
quality_data = quality_data.dropna(
    subset=["income_score"]
).copy()

# -------------------------------------------------------------------------
# Spearman Rank Correlation
# -------------------------------------------------------------------------

spearman_rho, p_value = stats.spearmanr(
    quality_data["income_score"],
    quality_data["rating"]
)

alpha = 0.05

# -------------------------------------------------------------------------
# Interpret correlation strength
# -------------------------------------------------------------------------

abs_rho = abs(spearman_rho)

if abs_rho < 0.20:
    strength = "Very weak"

elif abs_rho < 0.40:
    strength = "Weak"

elif abs_rho < 0.60:
    strength = "Moderate"

elif abs_rho < 0.80:
    strength = "Strong"

else:
    strength = "Very strong"


if spearman_rho > 0:
    direction = "positive"

elif spearman_rho < 0:
    direction = "negative"

else:
    direction = "no"


# -------------------------------------------------------------------------
# Hypothesis decision
# -------------------------------------------------------------------------

if p_value < alpha:

    decision = "Reject H₀"

    conclusion = (
        "There is statistically significant evidence at the 5% "
        "significance level of a monotonic relationship between "
        "Income Bracket and customer Rating."
    )

else:

    decision = "Fail to Reject H₀"

    conclusion = (
        "There is insufficient statistical evidence at the 5% "
        "significance level to conclude that Income Bracket and "
        "customer Rating have a monotonic relationship."
    )


# =============================================================================
# RESULTS
# =============================================================================

print("\n" + "=" * 70)
print("QUALITY EXPECTATIONS — SPEARMAN CORRELATION TEST")
print("=" * 70)

print("\nH₀: There is no statistically significant monotonic relationship")
print("    between Income Bracket and Rating.")

print("\nH₁: There is a statistically significant monotonic relationship")
print("    between Income Bracket and Rating.")

print(f"\nSignificance Level (α): {alpha}")

print(
    f"\nObservations Tested: "
    f"{len(quality_data):,}"
)

print(
    f"\nSpearman Correlation (ρ): "
    f"{spearman_rho:.4f}"
)

print(f"Correlation Strength: {strength}")
print(f"Correlation Direction: {direction}")

print(f"\nP-Value: {p_value:.6f}")

print(f"\nDecision: {decision}")

print("\nBusiness Interpretation:")
print(conclusion)


# =============================================================================
# AVERAGE RATING BY INCOME BRACKET
# =============================================================================

rating_summary = (
    quality_data
    .groupby("income_bracket", observed=True)
    .agg(
        Average_Rating=("rating", "mean"),
        Median_Rating=("rating", "median"),
        Rating_Count=("rating", "count")
    )
    .reset_index()
)

# Sort according to the defined income hierarchy
rating_summary["income_score"] = (
    rating_summary["income_bracket"]
    .map(income_order)
)

rating_summary = (
    rating_summary
    .sort_values("income_score")
    .drop(columns="income_score")
)

print("\n" + "=" * 70)
print("RATING BY INCOME BRACKET")
print("=" * 70)

display(
    rating_summary.round(2)
)


# =============================================================================
# VISUALIZATION — INCOME BRACKET VS RATING
# =============================================================================

plt.figure(figsize=(11, 6))

sns.boxplot(
    data=quality_data,
    x="income_bracket",
    y="rating",
    order=[
        x for x in income_order
        if x in quality_data["income_bracket"].unique()
    ]
)

plt.title(
    "Customer Ratings by Income Bracket",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Income Bracket")
plt.ylabel("Customer Rating")

plt.ylim(0.5, 5.5)

plt.tight_layout()
plt.show()
```

### Q7. Pricing & Cost Multicollinearity

##### Check whether pricing, cost, quantity, discount and profit variables have strong correlations.

```python
# Threshold:
# |Correlation| > 0.75 = Potential Multicollinearity Concern

# ------------------------------------------------------------
# 1. Merge product-level pricing/cost data with order data
# ------------------------------------------------------------
correlation_data = orders.merge(
    products[["product_id", "unit_cost", "retail_price"]],
    on="product_id",
    how="left"
)

# ------------------------------------------------------------
# 2. Select variables for correlation analysis
# ------------------------------------------------------------
corr_cols = [
    "unit_cost",
    "retail_price",
    "quantity",
    "discount_pct",
    "profit"
]

corr_data = correlation_data[corr_cols].copy()

# Convert columns to numeric
for col in corr_cols:
    corr_data[col] = pd.to_numeric(corr_data[col], errors="coerce")

# Remove rows with missing values
corr_data = corr_data.dropna()

print("Number of observations:", len(corr_data))

# ------------------------------------------------------------
# 3. Pearson correlation matrix
# ------------------------------------------------------------
corr_matrix = corr_data.corr(method="pearson")

print("\nCorrelation Matrix:")
display(corr_matrix.round(3))

# ------------------------------------------------------------
# 4. Visualize correlation matrix
# ------------------------------------------------------------
plt.figure(figsize=(9, 7))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="RdBu_r",
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.5
)

plt.title("Pricing & Cost Correlation Matrix")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 5. Identify high-correlation pairs
# ------------------------------------------------------------
threshold = 0.75

# Keep only upper triangle to avoid duplicate pairs
upper_triangle = corr_matrix.where(
    np.triu(
        np.ones(corr_matrix.shape),
        k=1
    ).astype(bool)
)

high_corr = (
    upper_triangle
    .stack()
    .reset_index()
)

high_corr.columns = [
    "Variable 1",
    "Variable 2",
    "Correlation"
]

high_corr["Absolute Correlation"] = (
    high_corr["Correlation"].abs()
)

high_corr = high_corr[
    high_corr["Absolute Correlation"] > threshold
].sort_values(
    "Absolute Correlation",
    ascending=False
)

# ------------------------------------------------------------
# 6. Display potential multicollinearity pairs
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("HIGH-CORRELATION PAIRS | THRESHOLD: |r| > 0.75")
print("=" * 70)

if high_corr.empty:
    print("No variable pairs exceed the |r| > 0.75 threshold.")
else:
    display(
        high_corr[
            ["Variable 1", "Variable 2", "Correlation"]
        ].round(3)
    )

# ------------------------------------------------------------
# 7. Business interpretation
# ------------------------------------------------------------
print("\nBusiness Interpretation:")

if high_corr.empty:
    print(
        "No strong linear relationships were identified among the "
        "selected variables using the |r| > 0.75 threshold."
    )
else:
    print(
        "The variable pairs listed above have strong linear "
        "relationships and should be reviewed for potential "
        "multicollinearity before being used together in a "
        "regression or predictive model."
    )

print(
    "\nNote: Correlation indicates linear association, not causation."
)
```
